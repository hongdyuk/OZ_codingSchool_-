[문제 3: 붕어빵 주문 처리]

손님의 주문을 처리하여 붕어빵 재고를 업데이트하는 기능을 구현하세요.

요구사항:
- 손님의 주문 내용을 기반으로 재고를 업데이트합니다.
- 재고가 부족할 경우, 손님에게 재고 부족을 알리고 재고를 감소시키지 않습니다.
- 재고가 충분할 경우, 주문한 만큼 재고를 감소시키고 판매를 완료합니다.
- 판매가 완료된 경우 판매된 붕어빵 맛과 개수를 출력하세요.

```python
inventory = {
    '팥붕어빵': 10,
    '슈크림붕어빵': 8,
    '초코붕어빵': 5
}

order = input(f'주문하실 메뉴와 개수를 입력해주세요!!ex) 팥붕어빵,2').split(",") 

flavor, count = order
count = int(count)

if flavor not in inventory:
    print(f"죄송합니다. {flavor} 맛은 판매하지 않습니다.")
else:
    if inventory[flavor] >= count:
        inventory[flavor] -= count
        print(f"{flavor} 맛 붕어빵 {count}개 판매 완료!")
    else:
        print(f"죄송합니다. {flavor} 맛 붕어빵의 재고가 부족합니다.")


print("\n업데이트된 재고 상태:")
print(inventory)
```

[문제 4: 여러 손님 주문 처리]

여러 손님들의 주문을 처리할 수 있도록 시스템을 확장하세요.

요구사항:
- 손님이 "종료"를 입력할 때까지 주문을 계속 받습니다.
- 각 주문마다 재고를 업데이트합니다.
- 현재 재고를 화면에 출력합니다.

```python
inventory = {
    '팥붕어빵': 10,
    '슈크림붕어빵': 8,
    '초코붕어빵': 5
}

while True:
    print("붕어빵 키오스크")
    print("현재 주문 가능한 붕어빵 재고량:")
    for flavor, count in inventory.items():
    # inventory에 있는 items을 불러와 keys와 value를 각각 flavor,count에 할당하면서
        print(f"{flavor}: {count}개")
        # for문을 이용해 3개의 요소를 출력하여 고객에게 재고현황을 보여줌
    
    print("\n어떤 붕어빵을 주문하시겠습니까? (종료하려면 'exit' 입력)")
    order = input().strip()
    # input으로 붕어빵의 맛을 받음
    
    if order.lower() == 'exit':
    # if문을 이용하여 종료 할수있도록 구현함 lower메서드는 대문자를 소문자로 변환해주기 위해 사용
        print("키오스크를 종료합니다.")
        break
        # exit를 입력하면 break되며 종료됌
    
    if order in inventory:
    # order가 inventory안에 있으면 if문 실행
        count = int(input(f"{order} 몇 개를 주문하시겠습니까? "))
        # input으로 개수를 입력받아 count에 할당함
        if inventory[order] >= count:
        # 만약 inventory에 있는 재고가 count보다 많거나 같으면 실행
            inventory[order] -= count
            # 선택한 맛의 재고에 입력받은 개수를 뺌
            print(f"{order} {count}개 주문 완료!")
        else:
        # 만약 inventory에 있는 재고가 count보다 적으면 실행
            print(f"죄송합니다. {order}의 재고가 부족합니다. 현재 재고: {inventory[order]}개")
    else:
        print(f"죄송합니다. {order}은(는) 판매하지 않는 붕어빵입니다.")


print("\n최종 재고 상태:")
for flavor, stock in inventory.items():
    print(f"{flavor}: {stock}개")
```

[문제 5: 관리자 모드]

관리자 모드를 통해 붕어빵 재고를 추가할 수 있는 기능을 구현하세요.

요구사항:
- 관리자는 특정 붕어빵의 재고를 추가할 수 있습니다.
- "종료"를 입력하면 관리자 모드를 종료합니다.
- 추가된 재고를 업데이트하고 현재 재고를 화면에 출력합니다.

```python
inventory = {
    '팥붕어빵': 10,
    '슈크림붕어빵': 15,
    '초코붕어빵': 8
}

while True:
    print("붕어빵 키오스크")
    print("현재 주문 가능한 붕어빵 재고량:")
    for flavor, count in inventory.items():
    # inventory에 있는 items을 불러와 keys와 value를 각각 flavor,count에 할당하면서
        print(f"{flavor}: {count}개")
        # for문을 이용해 3개의 요소를 출력하여 고객에게 재고현황을 보여줌

    admin_mode = input("\n관리자 모드에 접속하시겠습니까? (yes/no): ").lower()
    # input으로 답변을 admin_mode에 할당함
    if admin_mode == 'yes':
    # 할당받은 값이 yes면 실행
        while True:
        # 무한루프
            print("\n===== 관리자 모드 =====")
            print("1. 팥붕어빵 재고 추가")
            print("2. 슈크림붕어빵 재고 추가")
            print("3. 초코붕어빵 재고 추가")
            print("4. 종료")

            choice = input("번호를 선택하세요: ")
            # 변호를 입력받아 설정에 진입

            if choice == '1':
            # if문으로 입력받은 같이 1번이면 실행되도록 메뉴를 만듬
                amount = int(input("추가할 팥붕어빵의 수량을 입력하세요: "))
                inventory['팥붕어빵'] += amount
                # inventory의 재고에 수량을 입력받아 더함
                print(f"팥붕어빵 재고 {amount}개 추가 완료!")
                break
                # 관리자 모드 끝

            elif choice == '2':
                amount = int(input("추가할 슈크림붕어빵의 수량을 입력하세요: "))
                inventory['슈크림붕어빵'] += amount
                print(f"슈크림붕어빵 재고 {amount}개 추가 완료!")
                break

            elif choice == '3':
                amount = int(input("추가할 초코붕어빵의 수량을 입력하세요: "))
                inventory['초코붕어빵'] += amount
                print(f"초코붕어빵 재고 {amount}개 추가 완료!")
                break

            elif choice == '4':
                print("관리자 모드를 종료합니다.")
                break

            else:
                print("잘못된 입력입니다. 다시 입력해 주세요.")

    elif admin_mode == 'no':
    # 관리자 모드 입력창에 no라고 입력하면 일반 주문창으로 들어오게 만듬

        print("\n어떤 붕어빵을 주문하시겠습니까? (종료하려면 'exit' 입력)")
        order = input().strip()
        # input으로 붕어빵의 맛을 받음

        if order.lower() == 'exit':
        # if문을 이용하여 종료 할수있도록 구현함 lower메서드는 대문자를 소문자로 변환해주기 위해 사용
            print("키오스크를 종료합니다.")
            break
            # exit를 입력하면 break되며 종료됌
        
        if order in inventory:
        # order가 inventory안에 있으면 if문 실행
            count = int(input(f"{order} 몇 개를 주문하시겠습니까? "))
            # input으로 개수를 입력받아 count에 할당함
            if inventory[order] >= count:
            # 만약 inventory에 있는 재고가 count보다 많거나 같으면 실행
                inventory[order] -= count
                # 선택한 맛의 재고에 입력받은 개수를 뺌
                print(f"{order} {count}개 주문 완료!")
            else:
            # 만약 inventory에 있는 재고가 count보다 적으면 실행
                print(f"죄송합니다. {order}의 재고가 부족합니다. 현재 재고: {inventory[order]}개")
        else:
            print(f"죄송합니다. {order}은(는) 판매하지 않는 붕어빵입니다.")

    else:
        print("잘못된 입력입니다. 다시 입력해 주세요.")
```

```python
prices = {
    "팥붕어빵": 1000,
    "슈크림붕어빵": 1200,
    "초코붕어빵": 1500
}
# 각 붕어빵의 가격을 딕셔너리로 관리
sales_count = {
    "팥붕어빵": 0,
    "슈크림붕어빵": 0,
    "초코붕어빵": 0
}
# 판매된 붕어빵의 종류와 개수를 딕셔너리로 관리

total_revenue = 0
# 전체 매출을 저장할 변수로 0으로 설정

while True:
    print("붕어빵 키오스크")
    print("현재 주문 가능한 붕어빵 재고량:")
    for flavor, count in inventory.items():
    # inventory에 있는 items을 불러와 keys와 value를 각각 flavor,count에 할당하면서
        print(f"{flavor}: {count}개")
        # for문을 이용해 3개의 요소를 출력하여 고객에게 재고현황을 보여줌

    admin_mode = input("\n관리자 모드에 접속하시겠습니까? (yes/no): ").lower()
    # input으로 답변을 admin_mode에 할당함
    if admin_mode == 'yes':
    # 할당받은 값이 yes면 실행
        while True:
        # 무한루프
            print("\n===== 관리자 모드 =====")
            print("1. 팥붕어빵 재고 추가")
            print("2. 슈크림붕어빵 재고 추가")
            print("3. 초코붕어빵 재고 추가")
            print("4. 종료")

            choice = input("번호를 선택하세요: ")
            # 변호를 입력받아 설정에 진입

            if choice == '1':
            # if문으로 입력받은 같이 1번이면 실행되도록 메뉴를 만듬
                amount = int(input("추가할 팥붕어빵의 수량을 입력하세요: "))
                inventory['팥붕어빵'] += amount
                # inventory의 재고에 수량을 입력받아 더함
                print(f"팥붕어빵 재고 {amount}개 추가 완료!")
                break
                # 관리자 모드 끝

            elif choice == '2':
                amount = int(input("추가할 슈크림붕어빵의 수량을 입력하세요: "))
                inventory['슈크림붕어빵'] += amount
                print(f"슈크림붕어빵 재고 {amount}개 추가 완료!")
                break

            elif choice == '3':
                amount = int(input("추가할 초코붕어빵의 수량을 입력하세요: "))
                inventory['초코붕어빵'] += amount
                print(f"초코붕어빵 재고 {amount}개 추가 완료!")
                break

            elif choice == '4':
                print("관리자 모드를 종료합니다.")
                break

            else:
                print("잘못된 입력입니다. 다시 입력해 주세요.")

    elif admin_mode == 'no':
    # 관리자 모드 입력창에 no라고 입력하면 일반 주문창으로 들어오게 만듬

        print("\n어떤 붕어빵을 주문하시겠습니까? (종료하려면 'exit' 입력)")
        order = input().strip()
        # input으로 붕어빵의 맛을 받음

        if order.lower() == 'exit':
        # if문을 이용하여 종료 할수있도록 구현함 lower메서드는 대문자를 소문자로 변환해주기 위해 사용
            print("키오스크를 종료합니다.")
            break
            # exit를 입력하면 break되며 종료됌

        if order in inventory:
            # order가 inventory안에 있으면 if문 실행
            count = int(input(f"{order} 몇 개를 주문하시겠습니까? "))
            # input으로 개수를 입력받아 count에 할당함
            if inventory[order] >= count:
            # 만약 inventory에 있는 재고가 count보다 많거나 같으면 실행
                inventory[order] -= count
                # 선택한 맛의 재고에 입력받은 개수를 뺌
                sales_count[order] += count
                # 판매된 붕어빵의 개수에 주문한 개수를 더함
                total_revenue += prices[order] * count
                # 매출에 주문한 붕어빵의 가격에 개수를 곱하여 더함
                print(f"{order} {count}개 주문 완료!")
            else:
                print(f"죄송합니다. {order}의 재고가 부족합니다. 현재 재고: {inventory[order]}개")
        else:
            print(f"죄송합니다. {order}은(는) 판매하지 않는 붕어빵입니다.")

    else:
        print("잘못된 입력입니다. 다시 입력해 주세요.")
```
