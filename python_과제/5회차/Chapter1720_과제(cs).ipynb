객체지향에 대해서 설명해주세요

### 객체지향 프로그래밍(Object-Oriented Programming, OOP)은 <br>프로그램을 객체라는 단위로 구성하여 문제를 해결하는 프로그래밍 패러다임입니다. <br><br> OOP의 주요 개념은 다음과 같습니다:

#### 클래스와 객체:
##### 클래스: 
- 객체를 생성하기 위한 설계도입니다.
##### 객체: 
- 클래스의 인스턴스이며, 속성과 메서드를 가집니다.

#### 캡슐화(Encapsulation):

- 객체의 내부 상태(데이터)를 숨기고, 외부와의 상호작용을 제한합니다.
- 데이터를 보호하고, 코드의 무결성을 유지합니다.

#### 상속(Inheritance):

- 한 클래스가 다른 클래스의 속성과 메서드를 물려받는 것입니다.
- 코드 재사용성을 높이고, 계층 구조를 만듭니다.

#### 다형성(Polymorphism):

- 동일한 메서드나 연산자가 다양한 객체에서 다르게 동작하는 것입니다.
- 같은 인터페이스로 다른 객체들을 처리할 수 있습니다.

#### 추상화(Abstraction):

- 복잡한 시스템에서 중요한 부분만을 간추려 단순화하는 것입니다.
- 인터페이스와 추상 클래스를 통해 구현됩니다.

[문제3]

클래스 상속에 대해 설명해주세요

## 클래스 상속
- 기존 클래스의 속성과 메서드를 다른 클래스가 재사용할 수 있도록 해줌
- 부모클래스 : 상속을 제공하는 클래스, 기본 기능을 정의하고 여러 자식 클래스에서 공통된 특성을 상속받을 수 있다.
- 자식클래스 : 다른 클래스에서 정의된 속성과 메서드를 상속받는 클래스, 자식클래스는 부모클래스의 모든 기능을 그대로 사용하거나 필요에 따라 수정하여 사용가능
## 주요특징
- 코드 재사용성: 상속을 통해 기존의 코드를 재사용하여 새로운 기능을 빠르게 개발할 수 있습니다.
- 계층 구조: 클래스 간에 부모-자식 관계를 형성하여 계층 구조를 만들 수 있습니다. 이는 코드의 구조화와 관리를 용이하게 합니다.
- 메서드 오버라이딩: 자식 클래스에서 부모 클래스의 메서드를 재정의할 수 있습니다. 이를 통해 자식 클래스는 같은 이름의 메서드를 가지지만 다른 동작을 수행할 수 있습니다.
- 추가 기능 제공: 자식 클래스는 부모 클래스의 기능에 새로운 속성이나 메서드를 추가하여 확장할 수 있습니다.

클래스 다중 상속과 관련된 MRO에 대해 설명해주세요

## MRO(Method Resolution Order)
- 다중 상속을 사용할 때 메서드를 찾는 순서를 결정하는 알고리즘입니다. 
- 파이썬에서는 C3 선형화 알고리즘을 이용하여 다이아몬드 상속 문제를 해결하고, 각 클래스의 메서드를 적절한 순서로 호출할 수 있도록 합니다. 
- MRO는 클래스 정의 시에 파이썬 인터프리터에 의해 계산되며, __mro__ 속성을 통해 확인할 수 있습니다.

[문제5]

Self는 어떤 키워드 인가요?

## self
- 파이썬에서 인스턴스 메서드에서 첫 번째 매개변수로 사용되는 특별한 키워드입니다. 
- 이 키워드를 통해 메서드가 속한 인스턴스 자체에 접근할 수 있습니다. 
- self를 이용하여 인스턴스 변수를 설정하거나 인스턴스 메서드를 호출할 수 있습니다. 
- Java나 C++에서의 this와 유사한 역할을 합니다.

[문제6]

ZeroDivisionError,  IndexError 외에 3가지 이상의 파이썬 에러에 대해 설명해주세요

### TypeError (타입 에러):

- 함수나 연산에 부적절한 데이터 타입이 전달되었을 때 발생합니다. 
- 정수와 문자열을 더하는 연산에서 발생할 수 있습니다.

### NameError (이름 에러):

- 정의되지 않은 변수나 함수를 사용하려고 할 때 발생합니다. 
- 변수를 선언하지 않고 사용하거나 오타가 있는 변수명을 사용할 경우 발생할 수 있습니다.

### ValueError (값 에러):

- 함수에 전달된 인자의 값이 유효하지 않을 때 발생합니다. 
- 정수로 변환할 수 없는 문자열을 int() 함수에 전달하면 발생할 수 있습니다.

### KeyError (키 에러):

- 딕셔너리에서 존재하지 않는 키를 사용하여 요소에 접근하려고 할 때 발생합니다.

### AttributeError (속성 에러):

- 객체가 존재하지 않는 속성을 참조하려고 시도할 때 발생합니다. 
- 문자열 객체에 존재하지 않는 메서드를 호출하려고 할 때 발생할 수 있습니다.
