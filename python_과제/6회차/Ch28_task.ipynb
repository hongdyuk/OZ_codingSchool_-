사용자로부터 3개의 숫자를 입력받아, 이를 튜플로 만든 후 최소값과 최대값을 출력하는 함수 min_max를 작성하세요.

```python
def min_max(*values):
    min_value = min(values)
    max_value = max(values)

    for i in values:
        if i == min_value:
            print('최솟값 : ',i)
        elif i == max_value:
            print('최댓값 : ',i)
        else:
            print('사잇값 : ',i)


nums = list(map(int, input("원하시는 숫자를 입력해주세요").split(',')))
min_max(*nums)
```
