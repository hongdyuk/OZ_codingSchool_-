팩토리얼을 계산하는 재귀 함수 factorial을 작성하세요.

```python
def factorial(n):
    if n == 0:
    # 만약 n이 0이면 1로 반환
        return 1
        # 조기 리턴 1 
    else:
    # 만약 n이 0이 아니라면 아래를 실행
        return n * factorial(n-1)
        # n과 factorial()를 곱하는데 다시 상위 함수로 돌아가 다시 n-1값을 대입해 함수 실행 값이 0이 나올때까지 반복

factorial(5)
# factorial()에 5를 입력
```
