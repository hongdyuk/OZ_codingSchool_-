여러 숫자를 입력받아 그 합계를 반환하는 함수 sum_numbers를 작성하세요. 이 때, 가변 매개변수를 사용하세요.

```python
num = 0

def sum_numbers(*numbers):
    global num
    for n in numbers:
        num += n
    return num

numbers = list(map(int, input('합산할 수를 입력해주세요(ex. 1,2,3...)').split(',')))
result = sum_numbers(*numbers)
print(result)
```
