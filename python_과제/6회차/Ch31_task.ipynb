문자열 리스트를 받아, 각 문자열의 길이를 출력하는 함수 print_lengths를 작성하세요.

```python
def print_lengths(s):
# 입력받은 값으로 함수를 실행
    for i in s[0:]:
    # s에 담긴 시퀀스 순서대로 아래에 실행
        print(f'{i}의 문자개수는 {len(i)}개입니다.')
        # i로 s의 시퀀스 순서대로 입력 받으면 print()로 출력

str = list(input("아무말이나 입력하세요").split(','))
# input에 입력받은 값을 list로 형변환해주고 str변수에 할당
print_lengths(str)
# print_lengths()함수에 인자를 입력
```
